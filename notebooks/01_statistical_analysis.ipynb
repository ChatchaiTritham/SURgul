## 1.3. Hypothesis Testing

### Independent Samples T-Test

**Question:** Is there a significant difference in Body Mass Index (BMI) between patients who experienced an adverse event and those who did not?

- **Null Hypothesis (H0):** The mean BMI is the same for both groups.
- **Alternative Hypothesis (H1):** The mean BMI is different for the two groups.

```python
group_adverse_event = df[df['adverse_event'] == 1]['bmi']
group_no_adverse_event = df[df['adverse_event'] == 0]['bmi']

# Perform the t-test
t_stat, p_value = stats.ttest_ind(group_adverse_event, group_no_adverse_event, nan_policy='omit')

print(f"T-statistic: {t_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("\nConclusion: We reject the null hypothesis. There is a significant difference in BMI between the two groups.")
else:
    print("\nConclusion: We fail to reject the null hypothesis. There is no significant difference in BMI.")
```

### Chi-Squared Test of Independence

**Question:** Is there a statistically significant association between sex and the occurrence of an adverse event?

- **Null Hypothesis (H0):** Sex and adverse events are independent.
- **Alternative Hypothesis (H1):** Sex and adverse events are not independent.

```python
# Create a contingency table
contingency_table = pd.crosstab(df['sex'], df['adverse_event'])

print("Contingency Table:")
print(contingency_table)

# Perform the Chi-squared test
chi2_stat, p_value, dof, expected = stats.chi2_contingency(contingency_table)

print(f"\nChi-squared Statistic: {chi2_stat:.4f}")
print(f"P-value: {p_value:.4f}")

if p_value < 0.05:
    print("\nConclusion: We reject the null hypothesis. There is a significant association between sex and adverse events.")
else:
    print("\nConclusion: We fail to reject the null hypothesis. The two variables are independent.")
```
