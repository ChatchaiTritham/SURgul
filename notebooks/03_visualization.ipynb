## 9. Additional Analysis: Error Analysis

Detailed analysis of prediction errors.

```python
# Identify errors (excluding abstentions)
non_abstention = results_conservative[~results_conservative['is_abstention']].copy()
non_abstention['error'] = non_abstention['predicted'] != non_abstention['ground_truth']
non_abstention['error_magnitude'] = abs(non_abstention['predicted'] - non_abstention['ground_truth'])

# Error statistics
print("\n" + "="*60)
print("ERROR ANALYSIS")
print("="*60)
print(f"\nTotal non-abstention cases: {len(non_abstention)}")
print(f"Correct predictions: {(~non_abstention['error']).sum()}")
print(f"Incorrect predictions: {non_abstention['error'].sum()}")
print(f"Accuracy: {(~non_abstention['error']).mean():.3f}")
print(f"\nError magnitude distribution:")
print(non_abstention['error_magnitude'].value_counts().sort_index())
print("="*60)
```

```python
# Visualize error distribution
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Error rate by ground truth tier
error_by_tier = non_abstention.groupby('ground_truth')['error'].mean() * 100
axes[0].bar(range(1, 6), error_by_tier.values, color=sns.color_palette('colorblind', 5))
axes[0].set_xlabel('Ground Truth Risk Tier', fontsize=12)
axes[0].set_ylabel('Error Rate (%)', fontsize=12)
axes[0].set_title('Error Rate by Risk Tier', fontsize=14, fontweight='bold')
axes[0].set_xticks(range(1, 6))
axes[0].set_xticklabels(['R1', 'R2', 'R3', 'R4', 'R5'])
axes[0].grid(axis='y', alpha=0.3)

# Add values on bars
for i, v in enumerate(error_by_tier.values):
    axes[0].text(i+1, v + 0.5, f"{v:.1f}%", ha='center', va='bottom', fontweight='bold')

# Error magnitude distribution
error_magnitude_dist = non_abstention[non_abstention['error']]['error_magnitude'].value_counts().sort_index()
axes[1].bar(error_magnitude_dist.index, error_magnitude_dist.values, color='coral')
axes[1].set_xlabel('Error Magnitude (tiers)', fontsize=12)
axes[1].set_ylabel('Frequency', fontsize=12)
axes[1].set_title('Error Magnitude Distribution', fontsize=14, fontweight='bold')
axes[1].grid(axis='y', alpha=0.3)

# Add values on bars
for i, v in error_magnitude_dist.items():
    axes[1].text(i, v + 0.2, str(v), ha='center', va='bottom', fontweight='bold')

plt.tight_layout()
plt.savefig('../figures/error_analysis.png', dpi=300, bbox_inches='tight')
plt.show()

print("Figure saved: figures/error_analysis.png")
```

## 11. Figure Quality Check

```python
# Check figure specifications
from PIL import Image

print("\n" + "="*60)
print("FIGURE QUALITY CHECK")
print("="*60)

figures_dir = '../figures'
if os.path.exists(figures_dir):
    for filename in sorted(os.listdir(figures_dir)):
        if filename.endswith('.png'):
            filepath = os.path.join(figures_dir, filename)
            with Image.open(filepath) as img:
                width, height = img.size
                dpi = img.info.get('dpi', (None, None))[0]
                file_size = os.path.getsize(filepath) / 1024  # KB
                
                print(f"\n{filename}:")
                print(f"  Resolution: {width} x {height} pixels")
                print(f"  DPI: {dpi if dpi else 'Not specified'}")
                print(f"  File size: {file_size:.1f} KB")
                
                # Check if meets publication standards
                if dpi and dpi >= 300:
                    print(f"  โ“ Publication quality (โฅ300 DPI)")
                else:
                    print(f"  โ  May need higher DPI for print")

print("\n" + "="*60)
print("Quality check complete!")
print("="*60)
```
