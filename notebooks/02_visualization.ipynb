# 2. Visualization of Synthetic Clinical Data

This notebook provides scripts for visualizing the synthetic dataset. Good visualizations are key to understanding data distributions, relationships between variables, and communicating findings effectively. We will create several common plot types and save them to the `/visualizations` directory.

## 2.1. Imports and Data Generation

```python
import pandas as pd
import numpy as np
import sys
import os
import matplotlib.pyplot as plt
import seaborn as sns

# Add the 'src' directory to the system path
sys.path.append('../src')

from synthetic_data_generator import generate_syndx_like_data

# Generate a dataset
df = generate_syndx_like_data(n_samples=1000, noise_level=0.7)

# Ensure the output directory exists
output_dir = '../visualizations'
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

print("Dataset generated and output directory prepared.")
```

## 2.2. Correlation Heatmap

A heatmap is an excellent way to visualize the correlation matrix from the previous notebook.

```python
continuous_vars = ['age', 'bmi', 'systolic_bp', 'diastolic_bp', 'cholesterol']
correlation_matrix = df[continuous_vars].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Continuous Variables', fontsize=16)
plt.xticks(rotation=45)
plt.yticks(rotation=0)
plt.tight_layout()

# Save the figure
plt.savefig(os.path.join(output_dir, 'correlation_heatmap.png'), dpi=300)

plt.show()
```

## 2.3. Distribution Plots

### Box Plot for BMI by Adverse Event
This plot visually supports the t-test from the previous notebook, comparing the distributions of BMI for each group.

```python
plt.figure(figsize=(8, 6))
sns.boxplot(x='adverse_event', y='bmi', data=df, palette='Set2')
plt.title('BMI Distribution by Adverse Event', fontsize=14)
plt.xlabel('Adverse Event (0 = No, 1 = Yes)')
plt.ylabel('Body Mass Index (BMI)')
plt.tight_layout()

plt.savefig(os.path.join(output_dir, 'bmi_by_adverse_event_boxplot.png'), dpi=300)
plt.show()
```

### Histogram for Age
Shows the age distribution of the entire patient cohort.

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['age'], bins=20, kde=True, color='skyblue')
plt.title('Age Distribution of Patient Cohort', fontsize=14)
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.tight_layout()

plt.savefig(os.path.join(output_dir, 'age_distribution_histogram.png'), dpi=300)
plt.show()
```

## 2.4. Relationship and Categorical Plots

### Scatter Plot: BMI vs. Cholesterol
This helps to visualize the relationship between two continuous variables.

```python
plt.figure(figsize=(10, 7))
sns.scatterplot(x='bmi', y='cholesterol', hue='adverse_event', data=df, alpha=0.6, palette='viridis')
plt.title('BMI vs. Cholesterol, Colored by Adverse Event', fontsize=14)
plt.xlabel('Body Mass Index (BMI)')
plt.ylabel('Cholesterol')
plt.legend(title='Adverse Event')
plt.tight_layout()

plt.savefig(os.path.join(output_dir, 'bmi_vs_cholesterol_scatterplot.png'), dpi=300)
plt.show()
```

### Count Plot for Adverse Events by Sex
This plot supports the Chi-squared test by visualizing the frequency of adverse events in each sex.

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='sex', hue='adverse_event', data=df, palette='pastel')
plt.title('Count of Adverse Events by Sex', fontsize=14)
plt.xlabel('Sex (0 = Female, 1 = Male)')
plt.ylabel('Count')
plt.legend(title='Adverse Event', labels=['No', 'Yes'])
plt.tight_layout()

plt.savefig(os.path.join(output_dir, 'adverse_event_by_sex_countplot.png'), dpi=300)
plt.show()
```
